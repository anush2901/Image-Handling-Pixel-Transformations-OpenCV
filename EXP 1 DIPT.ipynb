10. Read in the image ('Apollo-11-launch.jpg').

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
text = "Apollo 11 Launch"

font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
thickness = 2

(text_width, text_height), _ = cv2.getTextSize(text, font, font_scale, thickness)

x = (img_apollo.shape[1] - text_width) // 2
y = img_apollo.shape[0] - 30

cv2.putText(img_apollo, text, (x, y), font, font_scale, (255, 255, 255), thickness)

plt.imshow(img_apollo[:, :, ::-1])
plt.show()
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
import cv2
import matplotlib.pyplot as plt

# Read boy.jpg
img_boy = cv2.imread('boy.jpg')

# Create contrast images
lower_contrast = cv2.convertScaleAbs(img_boy, alpha=0.5, beta=0)
higher_contrast = cv2.convertScaleAbs(img_boy, alpha=1.5, beta=0)

# Display all images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_boy[:, :, ::-1])
plt.title("Original")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(lower_contrast[:, :, ::-1])
plt.title("Lower Contrast")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(higher_contrast[:, :, ::-1])
plt.title("Higher Contrast")
plt.axis('off')

plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
# Split the image into B, G, R channels
b, g, r = cv2.split(img_boy)

# Display the channels
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(b, cmap='gray')
plt.title("Blue Channel")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='gray')
plt.title("Green Channel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(r, cmap='gray')
plt.title("Red Channel")
plt.axis('off')

plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
# Merge the R, G, B channels
merged = cv2.merge([b, g, r])

# Display Original and Merged images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_boy[:, :, ::-1])
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged[:, :, ::-1])
plt.title("Merged Image")
plt.axis('off')

plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
# Convert BGR image to HSV
hsv = cv2.cvtColor(img_boy, cv2.COLOR_BGR2HSV)

# Split into H, S, V channels
h, s, v = cv2.split(hsv)

# Display the channels
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(h, cmap='gray')
plt.title("Hue (H)")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(s, cmap='gray')
plt.title("Saturation (S)")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(v, cmap='gray')
plt.title("Value (V)")
plt.axis('off')

plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
# Merge H, S, V channels
merged_hsv = cv2.merge([h, s, v])

# Convert HSV back to BGR for displaying
merged = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2BGR)

# Display Original and Merged images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_boy[:, :, ::-1])
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged[:, :, ::-1])
plt.title("Merged HSV Image")
plt.axis('off')

plt.show()
```

```python
# Convert BGR to HSV
hsv = cv2.cvtColor(img_boy, cv2.COLOR_BGR2HSV)

# Split into H, S, V channels
h, s, v = cv2.split(hsv)

# Merge H, S, V channels back together
merged_hsv = cv2.merge([h, s, v])

# Convert merged HSV back to BGR
merged = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2BGR)

# Convert BGR images to RGB for Matplotlib
original_rgb = cv2.cvtColor(img_boy, cv2.COLOR_BGR2RGB)
merged_rgb = cv2.cvtColor(merged, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(h, cmap="hsv")
plt.title("H - Hue")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(s, cmap="gray")
plt.title("S - Saturation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(v, cmap="gray")
plt.title("V - Value")
plt.axis("off")

plt.tight_layout()
plt.show()
```
